```python
import pandas as pd
import numpy as np
from scipy import stats
from scipy.stats import chi2_contingency
from scipy.stats import ttest_ind
```

```python
df=pd.read_csv(r"Cleaned_Customer_Churn.csv")
```

```python
from scipy.stats import chi2_contingency

cat_cols = ['gender','Family_Status', 'SeniorCitizen', 'Contract', 'PaymentMethod', 'InternetService','Customer_Status']

for col in cat_cols:
    print(f"\n--- {col} vs Churn ---")
    
    contingency_table = pd.crosstab(df[col], df['Churn'])
    
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    
    # Cramér's V
    n = contingency_table.sum().sum()
    r, c = contingency_table.shape
    cramers_v = np.sqrt(chi2 / (n * (min(r, c) - 1)))
    
    print(f"Chi2: {chi2:.3f}")
    print(f"p-value: {p:.5f}")
    print(f"Cramér's V: {cramers_v:.3f}")
    
    # Interpretation
    if p < 0.05:
        print("Significant association")
    else:
        print("No significant association")
```

```python
# Q.) Is there any difference between Mean Monthly charges of customer who churn and who donot

Churn_no_MonthlyCharges = df.loc[df['Churn']==False,['MonthlyCharges']]
Churn_yes_MonthlyCharges = df.loc[df['Churn']==True,['MonthlyCharges']]

statistic, p = stats.ttest_ind(Churn_no_MonthlyCharges, Churn_yes_MonthlyCharges)

if p < 0.05:
    print(f"p vaule: {p} , there is a significant difference between Mean Monthly charges of customer who churn and who donot")
else:
    print(f"p vaule: {p}, there is not any differences between  between Mean Monthly charges of customer who churn and who donot")
```

```python
# Q.) Is there any difference in Tenure of customers who churn and who didnot churn

Churn_no_tenure=df.loc[df['Churn']==False,['tenure']]
Churn_yes_tenure = df.loc[df['Churn']==True,['tenure']]

statistic, p = stats.ttest_ind(Churn_no_tenure, Churn_yes_tenure )

if p < 0.05:
    print(f"p vaule: {p} , there is a significant difference between Mean Tenure of customers who churn and who donot")
else:
    print(f"p vaule: {p}, there is not any differences between between Mean Tenure of customers who churn and who donot")
```

```python
# Q.) Is there any difference between Total Charges of customer who churn and who donot

Churn_no_TotalCharges = df.loc[df['Churn']==False,['TotalCharges']]
Churn_yes_TotalCharges = df.loc[df['Churn']==True,['TotalCharges']]

statistic, p = stats.ttest_ind(Churn_no_TotalCharges, Churn_yes_TotalCharges)

if p < 0.05:
    print(f"p vaule: {p} , there is a significant difference between Mean Total charges of customer who churn and who donot")
else:
    print(f"p vaule: {p}, there is not any differences between  between Mean Total charges of customer who churn and who donot")
```
